## 1. Boolean Values and Comparisons

A Boolean value is either `True` or `False`. The Python type is `bool`. Comparison operators compare two values and return a Boolean result.

| Operator | Meaning | Example |
| --- | --- | --- |
| `==` | equal to | `score == 50` |
| `!=` | not equal to | `status != 'closed'` |
| `>` | greater than | `temperature > 30` |
| `<` | less than | `age < 18` |
| `>=` | greater than or equal to | `mark >= 50` |
| `<=` | less than or equal to | `speed <= 60` |

Use `==` to compare values. A single `=` assigns a value to a variable.


````python
# Comparison expressions return Boolean values.
# Set example values with different data types.
age = 19
score = 72
course_status = 'active'

# Evaluate each comparison and display its Boolean result.
print(age >= 18)
print(score == 50)
print(score != 0)
print(course_status == 'active')

# Confirm that a comparison produces a value of type bool.
print(type(age >= 18))
````

Python can chain comparisons. For example, `18 <= age < 65` checks both the lower and upper boundaries. This is equivalent to `age >= 18 and age < 65`.


````python
# Check whether a value is inside a range.
# Store the observed temperature.
temperature = 23

# Both boundaries are included in this chained comparison.
comfortable = 18 <= temperature <= 26

# Display the stored Boolean result.
print('Comfortable temperature:', comfortable)
````

## 2. Selection with `if`

Selection lets a program decide whether to run a block of code. An `if` statement runs its indented block only when the condition is `True`.

```python
if condition:
    statement
```

The colon ends the condition line. The four-space indentation shows which statements belong to the `if` block.


````python
# Run the indented statement when the condition is True.
# Set the value that will be tested.
weather = 'sunny'

# This message appears only when weather equals 'sunny'.
if weather == 'sunny':
    print('Take a hat and sunscreen.')

# This line is outside the if block, so it always runs.
print('Weather check complete.')
````

If the condition is `False` and there is no `else` block, Python skips the indented code and continues with the next statement.


````python
# This condition is False, so its indented message is skipped.
# Set a temperature below the condition's boundary.
temperature = 14

# The indented block runs only for temperatures above 30.
if temperature > 30:
    print('It is a hot day.')

# Program execution continues here after the condition is checked.
print('Temperature check complete.')
````

## 3. Two-Way Selection with `if` and `else`

Use `if` and `else` when exactly one of two paths should run. The `if` block runs when the condition is `True`; otherwise, the `else` block runs.

```python
if condition:
    statement_when_true
else:
    statement_when_false
```


````python
# Choose one of two messages.
# Store the condition as a Boolean value.
is_raining = True

# Exactly one of these two branches will run.
if is_raining:
    print('Take an umbrella.')
else:
    print('No umbrella is needed.')
````

A conditional expression is a compact way to choose one of two values. It is useful for a simple assignment, but a full `if`/`else` statement is clearer for several actions.


````python
# Choose one text value with a conditional expression.
# Set the mark that will be compared with the pass boundary.
mark = 64

# Select the first value when the condition is True; otherwise select the second.
result = 'Pass' if mark >= 50 else 'Not yet satisfactory'

# Display the selected text.
print(result)
````

## 4. Multi-Way Selection with `if`, `elif`, and `else`

Use `elif` (short for *else if*) when a decision has more than two possible paths. Python tests conditions from top to bottom and runs only the first matching block. The final `else` handles every remaining case.

```python
if first_condition:
    first_action
elif second_condition:
    second_action
else:
    default_action
```


````python
# Classify a mark with several ordered conditions.
# Set the mark that will be classified.
mark = 68

# Test boundaries from highest to lowest; only the first match runs.
if mark >= 80:
    grade = 'Distinction'
elif mark >= 65:
    grade = 'Credit'
elif mark >= 50:
    grade = 'Pass'
else:
    grade = 'Not yet satisfactory'

# Display the category selected by the conditional chain.
print('Grade:', grade)
````

Order matters. Put the most restrictive high boundary first. If `mark >= 50` appeared before `mark >= 80`, a mark of 85 would match the first condition and never reach the distinction condition.


````python
# Check ordered temperature boundaries from highest to lowest.
# Set the temperature that will be classified.
temperature = 31

# Stop at the first boundary that the temperature satisfies.
if temperature >= 35:
    description = 'hot'
elif temperature >= 25:
    description = 'warm'
elif temperature >= 15:
    description = 'mild'
else:
    description = 'cool'

# Display the description chosen by the conditional chain.
print('The weather is', description)
````

An `if` statement can be nested inside another `if` statement. Nesting is useful when the second test should happen only after the first test succeeds. Keep the indentation consistent.


````python
# Check age first, then identification.
# Store the two conditions required for entry.
age = 20
has_photo_id = True

# The inner identification check runs only when the age check succeeds.
if age >= 18:
    if has_photo_id:
        print('Entry approved.')
    else:
        print('Photo identification is required.')
else:
    print('Entry is not approved.')
````

## 5. Logical Operators: `and`, `or`, and `not`

Logical operators combine or reverse Boolean expressions. Python uses the lowercase keywords `and`, `or`, and `not`.

| Operator | Result is `True` when... | Example |
| --- | --- | --- |
| `and` | both expressions are true | `has_ticket and has_id` |
| `or` | at least one expression is true | `is_student or is_senior` |
| `not` | the expression is false | `not account_locked` |


### `and`

Use `and` when every condition must be `True`.


````python
# Both conditions must be True.
# Store the two entry requirements.
has_ticket = True
has_photo_id = True

# Logical AND requires both Boolean values to be True.
can_enter = has_ticket and has_photo_id

# Display the result of the combined condition.
print('Can enter:', can_enter)
````

### `or`

Use `or` when one or more conditions may be `True`.


````python
# At least one condition must be True.
# Store two independent discount conditions.
is_student = False
is_senior = True

# Logical OR accepts either condition or both conditions.
receives_discount = is_student or is_senior

# Display the result of the combined condition.
print('Receives discount:', receives_discount)
````

### `not`

Use `not` to reverse a Boolean value: `True` becomes `False`, and `False` becomes `True`.


````python
# Reverse the account_locked value.
# Store whether the account is currently locked.
account_locked = False

# Logical NOT changes False to True and True to False.
can_sign_in = not account_locked

# Display the reversed Boolean result.
print('Can sign in:', can_sign_in)
````

Logical operators can appear in one expression. Among comparison and logical operators, Python evaluates comparisons first, then `not`, then `and`, and finally `or`. Parentheses make the intended grouping easier to read.

| Priority | Operator |
| --- | --- |
| 1 (first) | comparisons such as `>=` and `==` |
| 2 | `not` |
| 3 | `and` |
| 4 (last) | `or` |


````python
# Combine comparisons and logical operators.
# Store the values used by the entry rule.
age = 17
has_photo_id = True
has_supervisor = True

# Approve an identified adult, or anyone accompanied by a supervisor.
can_enter = (age >= 18 and has_photo_id) or has_supervisor

# Display the final decision.
print('Can enter:', can_enter)
````

## 6. Boolean Expressions and Truth Tables

A Boolean expression evaluates to `True` or `False`. Complex rules become easier to read and test when they are translated into smaller, named conditions. A truth table can then compare every possible input combination and show whether a complex expression has a simpler equivalent.


Translate a written rule one phrase at a time, then give each important part a clear Boolean name.

| Written rule | Boolean formula |
| --- | --- |
| A and B must both be true | `A and B` |
| A or B may be true | `A or B` |
| A must be false | `not A` |
| A or B, but C must be false | `(A or B) and not C` |

Example rule: a learner is eligible for a fee concession when they are a student **or** hold a concession card, **and** they do **not** have overdue fees. The next cell separates this mixed rule into two named conditions before combining them.


````python
# Translate the fee-concession rule into named Boolean conditions.
# Store the three facts used by the rule.
is_student = True
has_concession_card = False
has_overdue_fees = False

# Translate each part of the written rule separately.
has_eligible_status = is_student or has_concession_card
fees_are_clear = not has_overdue_fees

# Combine the two named conditions with logical AND.
eligible_for_concession = has_eligible_status and fees_are_clear

# Display the smaller conditions and the final eligibility decision.
print('Has eligible status:', has_eligible_status)
print('Fees are clear:', fees_are_clear)
print('Eligible for concession:', eligible_for_concession)
````

### Simplifying a Boolean Expression with a Truth Table

A truth table lists every possible combination of Boolean inputs. It can simplify an expression by comparing the complex result with a shorter candidate expression.

Consider the complex expression `(A and B) or (A and not B)`. Compare its output with the simpler expression `A`.

Simplification process:

1. List every possible combination of `A` and `B`.
2. Calculate the intermediate parts and the complete expression.
3. Inspect the result pattern: it is always `False` when `A` is `False` and always `True` when `A` is `True`, regardless of `B`.
4. Propose `A` as the simpler expression and compare the two result columns.

| A | B | `A and B` | `A and not B` | Complex expression | `A` |
| --- | --- | --- | --- | --- | --- |
| `False` | `False` | `False` | `False` | `False` | `False` |
| `False` | `True` | `False` | `False` | `False` | `False` |
| `True` | `False` | `False` | `True` | `True` | `True` |
| `True` | `True` | `True` | `False` | `True` | `True` |

The final two columns match for every row. Therefore, `(A and B) or (A and not B)` is equivalent to `A`, so the complex expression can be simplified to `A`.


## 7. Bitwise Operators

Bitwise operators work with the individual binary digits, or *bits*, of integers. They are different from the logical operators `and`, `or`, and `not`.

`&`
: Bitwise AND. Example: `6 & 3`.

`|`
: Bitwise OR. Example: `6 | 3`.

`^`
: Bitwise XOR (exclusive OR). Example: `6 ^ 3`.

`~`
: Bitwise inversion. Example: `~6`.

`<<`
: Shift bits left. Example: `6 << 1`.

`>>`
: Shift bits right. Example: `6 >> 1`.

Python uses the single-character operators `&` and `|`. The operators `&&` and `||`, used by some other programming languages, are not valid Python syntax.


````python
# Compare integer values through their binary digits.
# Store decimal integers and annotate their four-bit representations.
left = 6   # binary 0110
right = 3  # binary 0011

# Display the original values as four binary digits.
print('left:        ', format(left, '04b'))
print('right:       ', format(right, '04b'))

# Apply AND, OR, and XOR to each matching pair of bits.
print('left & right:', format(left & right, '04b'), '=', left & right)
print('left | right:', format(left | right, '04b'), '=', left | right)
print('left ^ right:', format(left ^ right, '04b'), '=', left ^ right)

# Shift the bits one position left or right.
print('left << 1:   ', format(left << 1, '04b'), '=', left << 1)
print('left >> 1:   ', format(left >> 1, '04b'), '=', left >> 1)

# Invert every bit; for any integer n, ~n equals -(n + 1).
print('~left:', ~left)
````

### Using Bitwise Operators with Boolean Values

Because `bool` is a specialised integer type in Python, `&`, `|`, and `^` can also combine Boolean values.

`A & B`
: `True` when both `A` and `B` are true.

`A | B`
: `True` when at least one of `A` and `B` is true.

`A ^ B`
: `True` when exactly one of `A` and `B` is true.

For ordinary selection conditions, prefer `and`, `or`, and `not` because they clearly express logical intent and support short-circuit evaluation. Do not use `~` as Boolean NOT. Applying `~` directly to a Boolean value is deprecated in current Python versions; it historically produced `-2` for `~True`. Use `not True` to obtain `False`. If integer bit inversion is intended, convert explicitly, for example `~int(True)`.


````python
# Compare logical and bitwise operations on Boolean values.
# Store two Boolean inputs.
a = True
b = False

# Logical AND and bitwise AND give the same result for these bool values.
print('a and b:', a and b)
print('a & b:  ', a & b)

# Logical OR and bitwise OR also give the same bool result here.
print('a or b: ', a or b)
print('a | b:  ', a | b)

# XOR is True only when exactly one input is True.
print('a ^ b:  ', a ^ b)

# Use not for Boolean negation; convert to int before bitwise inversion.
print('not a:  ', not a)
print('~int(a):', ~int(a))
````

### Short-Circuit Evaluation

The logical operators `and` and `or` may skip the right-hand expression when the left-hand result already determines the answer. This is called *short-circuit evaluation*. Bitwise `&` and `|` always evaluate both operands. The next example uses simple, safe Boolean operands: both versions produce the same result, but their evaluation behaviour is different.

Bitwise operators also have higher precedence than comparisons. If they are intentionally used with Boolean comparisons, put each comparison in parentheses: `(age >= 18) & has_photo_id`. For ordinary scalar conditions, `age >= 18 and has_photo_id` is clearer.


````python
# Compare logical AND and bitwise AND with safe Boolean values.
# Store the values used by the entry conditions.
age = 16
has_photo_id = True

# Evaluate the age comparison once and store its Boolean result.
is_adult = age >= 18

# Logical AND can stop after is_adult because it is False.
logical_result = is_adult and has_photo_id

# Bitwise AND evaluates both Boolean operands.
bitwise_result = is_adult & has_photo_id

# The final values match even though the evaluation behaviour differs.
print('Logical AND result:', logical_result)
print('Bitwise AND result:', bitwise_result)
````

### Common Pitfalls

| Mistake | Problem | Correction |
| --- | --- | --- |
| `if score = 50:` | `=` assigns; it does not compare | `if score == 50:` |
| `if age >= 18 and <= 65:` | the second comparison has no value on its left | `if 18 <= age <= 65:` |
| `AND`, `OR`, `NOT` | Python keywords are lowercase | `and`, `or`, `not` |
| `condition1 && condition2` | `&&` is not Python syntax | `condition1 and condition2` |
| no indentation after `if` | Python cannot identify the selection block | indent the block by four spaces |
| `condition1 & condition2` for ordinary selection | `&` does not short-circuit | prefer `condition1 and condition2` |
| `~True` as Boolean NOT | direct `~` on `bool` is deprecated and historically produced `-2` | use `not True` |


````python
# Correct patterns for a range and a combined condition.
# Store the values required by the service rule.
age = 24
account_status = 'active'

# Check the inclusive age range first.
is_working_age = 18 <= age <= 65

# Require both a valid age and an active account.
can_use_service = is_working_age and account_status == 'active'

# Display the smaller condition and the final decision.
print('Working-age range:', is_working_age)
print('Can use service:', can_use_service)
````
