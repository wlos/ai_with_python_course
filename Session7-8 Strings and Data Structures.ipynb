## 1. Strings

A **string** is an ordered, immutable sequence of characters. Python provides a rich library of built-in methods for searching, transforming, and formatting text.

Key things to remember:

- Strings are **immutable** — every method returns a *new* string; the original is unchanged.
- Strings support **indexing** (`s[0]`) and **slicing** (`s[1:5]`).
- Indexing is **zero-based** and negative indices count from the end.


### 1.1 Indexing and slicing

````python
# Strings behave like sequences of characters
s = "Hello, World!"

print(len(s))        # total number of characters: 13
print(s[0])          # first character: H
print(s[-1])         # last character: !
print(s[7:12])       # slice from index 7 up to (not including) 12: World
print(s[::-1])       # reverse the string using a step of -1
````

### 1.2 Case methods

| Method | Description | Example |
|--------|-------------|--------|
| `.upper()` | All uppercase | `"hello"` → `"HELLO"` |
| `.lower()` | All lowercase | `"HELLO"` → `"hello"` |
| `.title()` | Title Case | `"hello world"` → `"Hello World"` |
| `.capitalize()` | First char upper, rest lower | `"hELLO"` → `"Hello"` |
| `.swapcase()` | Swap upper ↔ lower | `"Hello"` → `"hELLO"` |


````python
text = "python PROGRAMMING is Fun"

print(text.upper())       # PYTHON PROGRAMMING IS FUN
print(text.lower())       # python programming is fun
print(text.title())       # Python Programming Is Fun
print(text.capitalize())  # Python programming is fun
print(text.swapcase())    # PYTHON programming IS fUN
````

### 1.3 Search and test methods

**Everyday methods** — these cover the vast majority of real use cases:

| Method | Returns | Description |
|--------|---------|-------------|
| `.find(sub)` | `int` | Index of first occurrence, or `-1` if not found |
| `.count(sub)` | `int` | Number of non-overlapping occurrences |
| `.startswith(prefix)` | `bool` | Does the string start with this? |
| `.endswith(suffix)` | `bool` | Does the string end with this? |
| `.isdigit()` | `bool` | All characters are digits? |
| `.isalpha()` | `bool` | All characters are letters? |

**Less commonly needed:**

| Method | Returns | Description |
|--------|---------|-------------|
| `.rfind(sub)` | `int` | Index of *last* occurrence, or `-1` |
| `.index(sub)` | `int` | Like `find()` but raises `ValueError` if not found |
| `.isalnum()` | `bool` | All characters are letters or digits? |
| `.isspace()` | `bool` | All characters are whitespace? |


````python
s = "Hello, World! Hello, Python!"

print(s.find("Hello"))        # 0 — index of first match
print(s.count("Hello"))       # 2 — total occurrences
print(s.startswith("Hello"))  # True
print(s.endswith("!"))        # True
````

````python
# Testing string content
print("12345".isdigit())    # True — only digit characters
print("abc".isalpha())      # True — only letter characters
print("Hello!".isalpha())   # False — contains an exclamation mark
````

### 1.4 Modify and transform methods

**Everyday methods:**

| Method | Description |
|--------|-------------|
| `.strip()` | Remove leading/trailing whitespace |
| `.replace(old, new)` | Replace all occurrences of `old` with `new` |
| `.split(sep)` | Split into a list by separator (default: whitespace) |
| `.join(iterable)` | Join a list of strings using this string as separator |

**Less commonly needed:**

| Method | Description |
|--------|-------------|
| `.lstrip()` / `.rstrip()` | Remove from left/right side only |

````python
# strip removes surrounding whitespace; repr() reveals exact content
raw = "   Hello, World!   "
print(repr(raw.strip()))   # 'Hello, World!'
````

````python
# replace substitutes occurrences of one substring with another
msg = "I like cats. Cats are great."
print(msg.replace("cats", "dogs"))       # I like dogs. Cats are great.
print(msg.replace("cats", "dogs", 1))    # replace only the first occurrence
````

````python
# split breaks a string into a list; join is the inverse operation
csv_line = "Alice,30,London"
parts = csv_line.split(",")        # split on comma separator
print(parts)                        # ['Alice', '30', 'London']

joined = " | ".join(parts)         # join list items with " | " as separator
print(joined)                        # Alice | 30 | London
````

### 1.5 String formatting

Python provides three ways to embed values inside strings:

| Method | Syntax | Example |
|--------|--------|--------|
| f-strings (recommended) | `f"...{expr}..."` | `f"Hello, {name}!"` |
| `.format()` | `"...{}...".format(val)` | `"Hello, {}!".format(name)` |
| `%` operator (legacy) | `"...%s..." % val` | `"Hello, %s!" % name` |


````python
name = "Alice"
age = 30

# f-string (Python 3.6+) — expressions evaluate inside {}
print(f"My name is {name} and I am {age} years old.")

# .format() — placeholders filled left-to-right
print("My name is {} and I am {} years old.".format(name, age))

# % operator — older style; %s for string, %d for integer
print("My name is %s and I am %d years old." % (name, age))
````

### 1.6 f-string formatting mini-language

f-strings support a formatting specification after the colon: `{value:format_spec}`

| Spec | Meaning | Example | Result |
|------|---------|---------|--------|
| `:.2f` | 2 decimal places (float) | `f"{3.14159:.2f}"` | `"3.14"` |
| `:,` | Thousands separator | `f"{1000000:,}"` | `"1,000,000"` |
| `:>10` | Right-align in 10 chars | `f"{'hi':>10}"` | `"        hi"` |
| `:<10` | Left-align in 10 chars | `f"{'hi':<10}"` | `"hi        "` |
| `:^10` | Centre in 10 chars | `f"{'hi':^10}"` | `"    hi    "` |
| `:.2%` | Percentage with 2 decimals | `f"{0.856:.2%}"` | `"85.60%"` |
| `:08d` | Zero-padded integer | `f"{42:08d}"` | `"00000042"` |


````python
pi = 3.141592653589793
big_number = 1234567890
ratio = 0.8567

print(f"Pi to 4 decimals: {pi:.4f}")         # fixed-point with 4 decimal places
print(f"Big number: {big_number:,}")          # thousands separator
print(f"Ratio as percentage: {ratio:.2%}")    # percentage with 2 decimal places
print(f"Zero-padded: {42:08d}")               # integer padded to width 8 with zeros
````

````python
# Practical example: formatted product table
products = [("Apple", 1.20), ("Banana", 0.50), ("Cherry", 3.75)]

print(f"{'Product':<12} {'Price':>8}")   # header; < is left-align, > is right-align
print("-" * 21)
for product, price in products:
    print(f"{product:<12} ${price:>7.2f}")  # name left-aligned, price right-aligned
````

### 2.1 List

A **list** is an ordered, mutable collection — the most commonly used data structure in Python.

- Created with square brackets `[]`
- Can contain mixed types
- Supports indexing, slicing, and many built-in methods


````python
# Creating lists with different element types
numbers = [10, 20, 30, 40, 50]
mixed   = [1, "hello", 3.14, True]
empty   = []

print("numbers:", numbers)
print("mixed:", mixed)
````

````python
# Indexing (zero-based) and slicing
print("First element:", numbers[0])     # 10
print("Last element:",  numbers[-1])     # 50
print("Slice [1:4]:",   numbers[1:4])    # [20, 30, 40]
print("Every 2nd:",     numbers[::2])    # [10, 30, 50]
````

````python
# Lists are MUTABLE — elements can be changed after creation
numbers[0] = 99
print("After changing first element:", numbers)

# append adds a single item to the end of the list
numbers.append(60)
print("After append(60):", numbers)

# insert places an item at a specific index; existing items shift right
numbers.insert(1, 15)
print("After insert(1, 15):", numbers)

# remove deletes the first occurrence of the specified value
numbers.remove(30)
print("After remove(30):", numbers)

# pop removes and returns the last item (or item at a given index)
popped = numbers.pop()
print("Popped:", popped, "| Remaining:", numbers)

# sort reorders elements in ascending order by default
numbers.sort()
print("Sorted:", numbers)
````

#### List methods

**Everyday methods** — add, remove, and query:

| Method | Description | Returns |
|--------|-------------|--------|
| `append(x)` | Add item `x` to the end | `None` (in-place) |
| `insert(i, x)` | Insert item `x` at index `i` | `None` (in-place) |
| `remove(x)` | Remove first occurrence of `x` | `None` (in-place) |
| `pop(i=-1)` | Remove and return item at index `i` | The removed item |
| `sort(reverse=False)` | Sort the list in place | `None` (in-place) |
| `index(x)` | Index of first occurrence of `x` | `int` |
| `count(x)` | Count occurrences of `x` | `int` |

**Less commonly needed:**

| Method | Description | Returns |
|--------|-------------|--------|
| `extend(iterable)` | Append all items from an iterable | `None` (in-place) |
| `reverse()` | Reverse the list in place | `None` (in-place) |
| `copy()` | Return a shallow copy | New `list` |
| `clear()` | Remove all items | `None` (in-place) |


#### Operators on lists

| Operator | Meaning | Example |
|----------|---------|--------|
| `+` | Concatenation — combine two lists into a new one | `[1, 2] + [3, 4]` → `[1, 2, 3, 4]` |
| `*` | Repetition — repeat a list *n* times | `[0] * 3` → `[0, 0, 0]` |
| `==` / `!=` | Equality — compare element by element | `[1, 2] == [1, 2]` → `True` |
| `<`, `>`, `<=`, `>=` | Lexicographic comparison — like dictionary word order | `[1, 2] < [1, 3]` → `True` |


````python
a = [1, 2, 3]
b = [4, 5, 6]

# + creates a brand-new list from two existing ones
print("a + b =", a + b)           # [1, 2, 3, 4, 5, 6]

# * repeats the list the specified number of times
print("[0] * 4 =", [0] * 4)        # [0, 0, 0, 0]

# == compares element by element; order matters
print("[1, 2] == [1, 2]:", [1, 2] == [1, 2])   # True
print("[1, 2] == [2, 1]:", [1, 2] == [2, 1])   # False — different order

# < compares element by element from left to right (lexicographic)
print("[1, 2, 3] < [1, 2, 4]:", [1, 2, 3] < [1, 2, 4])  # True — 3 < 4 at index 2
````

#### List comprehensions

A **list comprehension** creates a list in a single expression:

```python
[expression for item in iterable if condition]
```


````python
# Squares of even numbers from 0 to 9
squares = [x ** 2 for x in range(10) if x % 2 == 0]
print("Even squares:", squares)  # [0, 4, 16, 36, 64]

# Dict comprehension maps each word to its character count
words = ["apple", "banana", "cherry"]
word_lengths = {w: len(w) for w in words}
print("Word lengths:", word_lengths)
````

### 2.2 Tuple

A **tuple** is like a list, except it is **immutable** — once created, it cannot be changed.

- Created with parentheses `()` (or just commas)
- Use tuples when data should not be modified (coordinates, RGB values, database rows)
- Faster than lists for read-only operations
- Can be used as dictionary keys (unlike lists)


````python
# Creating tuples
coordinates = (10.5, 20.3)
rgb_red = (255, 0, 0)
single  = (42,)          # trailing comma is required for single-element tuples

print("Coordinates:", coordinates)
print("RGB Red:", rgb_red)
print("Single:", single, "| Type:", type(single))
````

````python
# Tuples support indexing and slicing, just like lists
print("First coordinate:", coordinates[0])
print("Slice:", rgb_red[0:2])

# Tuple unpacking assigns each element to a separate variable
x, y = coordinates
print(f"x = {x}, y = {y}")

r, g, b = rgb_red
print(f"R={r}, G={g}, B={b}")

# Swapping two variables using tuple unpacking
a, b = 1, 2
a, b = b, a
print(f"After swap: a={a}, b={b}")
````

````python
# Tuples are IMMUTABLE — attempting to change an element raises TypeError
# Uncomment the line below to see the error:
# coordinates[0] = 99
````

#### Tuple methods

Because tuples are **immutable**, they have only two built-in methods:

| Method | Description | Returns |
|--------|-------------|--------|
| `count(x)` | Count occurrences of `x` | `int` |
| `index(x)` | Index of first occurrence of `x` | `int` |

For everything else (sorting, filtering, etc.), convert to a `list` first.


````python
colours = ("red", "green", "blue", "red", "green", "red")

# count — how many times 'red' appears
print("count('red'):", colours.count("red"))    # 3

# index — position of first occurrence
print("index('green'):", colours.index("green"))  # 1
````

### 2.3 Set

A **set** is an **unordered** collection of **unique** elements.

- Created with curly braces `{}` or `set()`
- Automatically removes duplicates
- Very fast membership testing (`in`)
- Supports mathematical set operations (union, intersection, difference)


````python
# Duplicates are automatically removed when creating a set
numbers = {1, 2, 2, 3, 3, 3, 4}
print("Unique numbers:", numbers)  # {1, 2, 3, 4}

# Converting a list to a set is a quick way to de-duplicate
grades = ["A", "B", "A", "C", "B", "A"]
unique_grades = set(grades)
print("Unique grades:", unique_grades)
````

#### Set methods

**Everyday methods** — add, remove, and query:

| Method | Description | Returns |
|--------|-------------|--------|
| `add(x)` | Add element `x` | `None` |
| `discard(x)` | Remove `x` — **no error** if absent | `None` |
| `remove(x)` | Remove `x` — **raises `KeyError`** if absent | `None` |
| `union(other)` | All elements from both sets | New `set` |
| `intersection(other)` | Elements in **both** sets | New `set` |
| `difference(other)` | Elements in `self` but not `other` | New `set` |

**Less commonly needed:**

| Method | Description | Returns |
|--------|-------------|--------|
| `symmetric_difference(other)` | Elements in one but not both | New `set` |
| `update(other)` | Add all elements from `other` (in-place union) | `None` |
| `intersection_update(other)` | Keep only common elements (in-place) | `None` |
| `issubset(other)` | Is every element of `self` also in `other`? | `bool` |
| `issuperset(other)` | Does `self` contain every element of `other`? | `bool` |
| `isdisjoint(other)` | Do the two sets share no elements? | `bool` |
| `pop()` | Remove and return an arbitrary element | Element |
| `clear()` | Remove all elements | `None` |
| `copy()` | Shallow copy | New `set` |


````python
colours = {"red", "green", "blue"}

# add inserts one new element (no-op if element already present)
colours.add("yellow")
print("After add:", colours)

# discard removes an element without raising an error if it is absent
colours.discard("green")
print("After discard:", colours)

# Sets are UNORDERED — indexing raises a TypeError
# colours[0]  # TypeError: 'set' object is not subscriptable
````

#### Operators on sets

Sets have their own mathematical operators. The comparison operators `<`, `<=`, `>`, `>=` test **subset and superset** relationships — not size or element values.

| Operator | Meaning | Example |
|----------|---------|--------|
| `\|` | Union — all elements from both sets | `{1,2} \| {2,3}` → `{1,2,3}` |
| `&` | Intersection — elements in both | `{1,2} & {2,3}` → `{2}` |
| `-` | Difference — in left but not right | `{1,2,3} - {2,3}` → `{1}` |
| `^` | Symmetric difference — in one but not both | `{1,2} ^ {2,3}` → `{1,3}` |
| `==` / `!=` | Equal sets (order doesn't matter) | `{1,2} == {2,1}` → `True` |
| `<` | Proper subset | `{1,2} < {1,2,3}` → `True` |
| `<=` | Subset (or equal) | `{1,2} <= {1,2}` → `True` |
| `>` | Proper superset | `{1,2,3} > {1,2}` → `True` |
| `>=` | Superset (or equal) | `{1,2} >= {1,2}` → `True` |


````python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("A | B (union):",               A | B)   # all elements from both
print("A & B (intersection):",         A & B)   # elements common to both
print("A - B (difference):",           A - B)   # in A but not in B
print("A ^ B (symmetric diff):",       A ^ B)   # in A or B, but not both

# Comparison operators test subset/superset, not element values
print()
print("{1,2} < {1,2,3}:", {1, 2} < {1, 2, 3})   # True — proper subset
print("{1,2} <= {1,2}:  ", {1, 2} <= {1, 2})      # True — subset (equal sets)
print("{1,2,3} > {1,2}: ", {1, 2, 3} > {1, 2})   # True — proper superset
print("{1,2} < {3,4}:   ", {1, 2} < {3, 4})       # False — neither is a subset
````

### 2.4 Dictionary

A **dictionary** stores data as **key–value pairs**. Think of it like a real dictionary: you look up a *word* (key) to find its *definition* (value).

- Created with curly braces `{key: value}` or `dict()`
- Keys must be unique and **immutable** (strings, numbers, tuples)
- Values can be any type
- Very fast lookup by key
- Maintains insertion order (Python 3.7+)


````python
# Creating a dictionary with string keys
student = {
    "name": "Alice",
    "age": 22,
    "courses": ["Maths", "Physics", "CS"],
    "graduated": False
}

print(student)
````

````python
# Access a value by its key using square bracket notation
print("Name:", student["name"])
print("Age:", student["age"])

# .get() returns None (or a custom default) instead of raising KeyError
print("GPA:", student.get("gpa"))          # None — key not present
print("GPA:", student.get("gpa", "N/A"))   # 'N/A' as default value
````

````python
# Dictionaries are MUTABLE — add, update, or delete keys after creation
student["age"] = 23              # update an existing key
student["gpa"] = 3.8             # add a new key-value pair
del student["graduated"]         # delete a key

print("Updated student:", student)
````

````python
# .keys(), .values(), and .items() give live views over the dictionary
print("Keys:",   list(student.keys()))
print("Values:", list(student.values()))
print("Items:",  list(student.items()))   # list of (key, value) tuples

# Iterating over key-value pairs is the standard pattern
for key, value in student.items():
    print(f"  {key}: {value}")
````

#### Dictionary methods

**Everyday methods** — access and modify:

| Method | Description | Returns |
|--------|-------------|--------|
| `keys()` | All keys | `dict_keys` view |
| `values()` | All values | `dict_values` view |
| `items()` | All `(key, value)` pairs | `dict_items` view |
| `get(key, default)` | Value for `key`, or `default` if missing | Value or default |
| `update(other)` | Merge `other` into this dict (overwrites existing keys) | `None` |
| `pop(key, default)` | Remove key and return its value | Value |

**Less commonly needed:**

| Method | Description | Returns |
|--------|-------------|--------|
| `setdefault(key, default)` | Like `get()`, but also inserts the default if missing | Value |
| `popitem()` | Remove and return the last inserted `(key, value)` pair | `tuple` |
| `clear()` | Remove all items | `None` |
| `copy()` | Shallow copy | New `dict` |
| `fromkeys(keys, value)` | Create a new dict from keys with a default value | New `dict` |


#### Operators on dictionaries

| Operator | Meaning | Note |
|----------|---------|------|
| `==` / `!=` | Equal — same key-value pairs (order doesn't matter) | Works on all Python versions |
| `\|` | Merge — combine two dicts into a new one (right-side wins on duplicate keys) | Python 3.9+ |
| `\|=` | In-place merge | Python 3.9+ |

> `+`, `-`, `<`, `>` are **not** supported on dictionaries.


````python
# == checks that two dicts have identical key-value pairs (insertion order is ignored)
d1 = {"a": 1, "b": 2}
d2 = {"b": 2, "a": 1}
print("d1 == d2:", d1 == d2)   # True — same pairs regardless of order

# | merges two dicts into a new one (Python 3.9+)
defaults = {"theme": "light", "language": "en", "font_size": 14}
user_prefs = {"theme": "dark", "font_size": 16}

# right-side values win when the same key appears in both dicts
config = defaults | user_prefs
print("Merged config:", config)
# {'theme': 'dark', 'language': 'en', 'font_size': 16}
````

#### What can be a dictionary key?

Keys must be **hashable** — they must be immutable so their value cannot change after creation.

| Type | Valid key? | Why? |
|------|:---:|---|
| `str` | ✅ | Strings are immutable |
| `int`, `float` | ✅ | Numbers are immutable |
| `tuple` | ✅\* | Tuples are immutable (if all elements are too) |
| `list` | ❌ | Lists are mutable |
| `dict` | ❌ | Dictionaries are mutable |
| `set` | ❌ | Sets are mutable |

\* *Only if all elements inside the tuple are also immutable.*


````python
# Tuples of immutable values make valid dictionary keys
locations = {
    (0, 0): "origin",
    (1, 0): "east",
    (0, 1): "north",
}
print(locations[(0, 0)])   # origin

# Lists cannot be keys — this raises TypeError: unhashable type: 'list'
# bad = {[1, 2]: "nope"}
````

## 3. Converting Between Data Structures

Python's constructor functions `list()`, `tuple()`, `set()`, and `dict()` convert between collection types — this is called **type casting**.

Common reasons to cast:

- **De-duplicate** a list → convert to `set`
- **Sort** an unordered set → convert to a sorted `list`
- **Make a list immutable** → convert to `tuple`
- **Build a dictionary** from a list of pairs → use `dict()`

| From ↓ / To → | `list()` | `tuple()` | `set()` | `dict()` |
|:--------------:|:--------:|:---------:|:-------:|:--------:|
| **List** | — | ✅ | ✅ (removes duplicates) | ✅\* |
| **Tuple** | ✅ | — | ✅ (removes duplicates) | ✅\* |
| **Set** | ✅ (arbitrary order) | ✅ | — | ❌ |
| **Dict** | ✅ (keys only) | ✅ (keys only) | ✅ (keys only) | — |
| **Range** | ✅ | ✅ | ✅ | ❌ |
| **String** | ✅ (list of chars) | ✅ | ✅ (unique chars) | ❌ |

\* *Converting to `dict` requires the source to contain 2-element pairs.*


````python
# List ↔ Tuple ↔ Set conversions
fruits_list  = ["apple", "banana", "cherry"]
fruits_tuple = tuple(fruits_list)              # list → tuple
print("list → tuple:", fruits_tuple)

numbers      = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]  # list with duplicates
unique       = set(numbers)                      # list → set removes duplicates
print("list → set:", unique)

sorted_unique = sorted(unique)                   # sorted() always returns a list
print("set → sorted list:", sorted_unique)
````

````python
# Building a dictionary from two parallel lists using zip()
keys   = ["name", "age", "city"]
values = ["Bob", 30, "Paris"]

# zip pairs corresponding elements; dict converts the pairs to key-value entries
person = dict(zip(keys, values))
print("zip → dict:", person)   # {'name': 'Bob', 'age': 30, 'city': 'Paris'}
````

````python
# Strings are iterable — each character becomes an element
word = "hello"
print("str → list:", list(word))    # ['h', 'e', 'l', 'l', 'o']
print("str → set: ", set(word))     # {'h', 'e', 'l', 'o'} — duplicates removed

# join rebuilds a string from an iterable of characters
chars = ["P", "y", "t", "h", "o", "n"]
print("list → str:", "".join(chars))   # Python
````

````python
# Common patterns using chained conversions
data = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]

# Pattern 1: remove duplicates and sort
unique_sorted = sorted(set(data))
print("Unique and sorted:", unique_sorted)

# Pattern 2: remove duplicates while preserving insertion order
# dict.fromkeys() keeps only the first occurrence of each value
unique_ordered = list(dict.fromkeys(data))
print("Unique (original order):", unique_ordered)
````

## 4. Case Study: Student Results Analyser

This case study combines strings, lists, sets, and dictionaries to build a small student results analyser. Follow each step and observe how the different collection types work together.


````python
# Dictionary mapping each student name to a list of scores
class_grades = {
    "Alice": [85, 92, 78, 90],
    "Bob":   [70, 65, 80, 72],
    "Cathy": [95, 98, 92, 97],
}

print(f"{'Name':<8} {'Average':>8} {'Grade':>6}")
print("-" * 24)

for name, scores in class_grades.items():
    average = sum(scores) / len(scores)   # compute mean score

    # assign a letter grade based on the average
    if average >= 90:
        grade = "A"
    elif average >= 80:
        grade = "B"
    elif average >= 70:
        grade = "C"
    else:
        grade = "F"

    print(f"{name:<8} {average:>8.1f} {grade:>6}")
````

````python
# Membership test to look up a specific student
query = "Diana"
if query in class_grades:
    print(f"{query}'s scores: {class_grades[query]}")
else:
    print(f"{query} is not in the class.")

# Collect all unique scores using a set
all_scores    = [score for scores in class_grades.values() for score in scores]
unique_scores = set(all_scores)   # remove duplicates
print(f"\nAll unique scores: {sorted(unique_scores)}")
print(f"Total unique scores: {len(unique_scores)}")
````
