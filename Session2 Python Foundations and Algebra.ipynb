## 1. Variables and Data Types

A **variable** is a name that refers to a Python object. Assignment uses `=`:

```python
variable_name = value
```

Python is dynamically typed. A name does not have one permanent type; it can be reassigned to an object of another type. The object still has a definite type, which we can inspect with `type()`.

| Type | Purpose | Example |
| --- | --- | --- |
| `int` | whole numbers | `42` |
| `float` | approximate real numbers | `3.5` |
| `bool` | logical values | `True` |
| `str` | text | `'Perth'` |


````python
# Create variables containing common scalar data types.
student_count = 24
room_temperature = 21.5
course_code = 'ICTPRG'
class_is_full = False

for name, value in [
    ('student_count', student_count),
    ('room_temperature', room_temperature),
    ('course_code', course_code),
    ('class_is_full', class_is_full),
]:
    print(f'{name:18} value={value!r:10} type={type(value).__name__}')
````

### Memory size of different data types

Every Python object requires memory for its value and for information used internally by Python. `sys.getsizeof()` reports the **shallow size** of one object in bytes.

The exact results depend on the Python version, operating system, and computer architecture. A longer string usually requires more memory than a shorter string, and a very large integer can require more memory than a small integer.


````python
import sys

# Compare the shallow memory size of the values created above.
sample_values = [
    ('int', student_count),
    ('float', room_temperature),
    ('bool', class_is_full),
    ('str', course_code),
]

print(f"{'Data type':10} {'Value':12} {'Memory size'}")
print('-' * 38)

for data_type, value in sample_values:
    print(f'{data_type:10} {value!r:12} {sys.getsizeof(value):>4} bytes')
````

`sys.getsizeof()` measures only the object passed to it. It does not add the sizes of other objects that may be referenced by that object.


````python
# Compare values of the same type but different sizes.
size_examples = [
    ('small integer', 10),
    ('large integer', 10**100),
    ('short string', 'AI'),
    ('long string', 'Artificial Intelligence with Python'),
]

for description, value in size_examples:
    print(f'{description:16} {sys.getsizeof(value):>4} bytes')
````

### Variable name composition and requirements

A Python variable name is an **identifier**. It must follow these rules:

| Rule | Valid example | Invalid example |
| --- | --- | --- |
| use letters, digits, and underscores only | `student_2` | `student-2` |
| begin with a letter or underscore, not a digit | `_score` | `2nd_score` |
| do not use spaces | `total_cost` | `total cost` |
| do not use a Python keyword | `class_name` | `class` |
| remember that names are case-sensitive | `score` and `Score` are different | — |

Unicode letters are permitted, but simple English names are generally easier to type, share, and maintain. Choose meaningful names and normally use lowercase `snake_case` for variables.


````python
import keyword

# Check whether candidate text can be used as a Python variable name.
candidate_names = [
    'student_count',
    '_temporary_value',
    'student2',
    '2nd_student',
    'total-cost',
    'class',
]

for candidate in candidate_names:
    is_valid = candidate.isidentifier() and not keyword.iskeyword(candidate)
    print(f'{candidate:18} valid={is_valid}')
````

### Type conversion

Use constructor functions such as `int()`, `float()`, and `str()` for explicit conversion. Conversion is only valid when the source value has a compatible form.


````python
# Convert compatible values explicitly.
quantity_text = '12'
quantity = int(quantity_text)
unit_price = 4.75
total_cost = quantity * unit_price

print(quantity, type(quantity))
print(total_cost, type(total_cost))
print('Total: $' + str(total_cost))
````

> **Common pitfall:** input from a keyboard or file may be text even when it looks numeric. `'12' + '3'` gives `'123'`, while `int('12') + int('3')` gives `15`.


````python
# Compare text concatenation with integer addition.
print('12' + '3')
print(int('12') + int('3'))
````

## 2. Binary, Decimal, and Hexadecimal Representation

A number's value is independent of how it is written.

| Base | Name | Digits | Example for decimal 26 |
| ---: | --- | --- | --- |
| 2 | binary | `0`, `1` | `11010₂` |
| 10 | decimal | `0`–`9` | `26₁₀` |
| 16 | hexadecimal | `0`–`9`, `A`–`F` | `1A₁₆` |

Python integer literals use the prefixes `0b` for binary and `0x` for hexadecimal.


````python
# These literals represent the same integer value.
decimal_value = 26
binary_value = 0b11010
hexadecimal_value = 0x1A

print(decimal_value, binary_value, hexadecimal_value)
print(decimal_value == binary_value == hexadecimal_value)
````

### Convert an integer to text in another base

Use `bin()` and `hex()` when prefixes are useful. Use `format()` for digits without prefixes or with padding.


````python
value = 45

print('Binary with prefix:', bin(value))
print('Hex with prefix:', hex(value))
print('8-bit binary:', format(value, '08b'))
print('Two-digit hexadecimal:', format(value, '02X'))
````

### Convert text to an integer

`int(text, base)` interprets a string using the supplied base. The returned object is an integer; decimal is only its usual display format.


````python
from_binary = int('101101', 2)
from_hexadecimal = int('2D', 16)

print(from_binary)
print(from_hexadecimal)
print(from_binary == from_hexadecimal)
````

### Place value

Binary `101101₂` can be expanded with powers of 2:

`1×2⁵ + 0×2⁴ + 1×2³ + 1×2² + 0×2¹ + 1×2⁰ = 45`

Hexadecimal `2D₁₆` uses `D = 13`:

`2×16¹ + 13×16⁰ = 45`


````python
binary_digits = '101101'
place_values = [
    int(digit) * 2**position
    for position, digit in enumerate(reversed(binary_digits))
]

print('Place values:', place_values)
print('Total:', sum(place_values))
````

Hexadecimal is compact and maps neatly to binary: one hexadecimal digit represents four bits.

| Hex | Binary | Hex | Binary |
| --- | --- | --- | --- |
| `0` | `0000` | `8` | `1000` |
| `1` | `0001` | `9` | `1001` |
| `2` | `0010` | `A` | `1010` |
| `3` | `0011` | `B` | `1011` |
| `4` | `0100` | `C` | `1100` |
| `5` | `0101` | `D` | `1101` |
| `6` | `0110` | `E` | `1110` |
| `7` | `0111` | `F` | `1111` |


````python
# Group binary digits in fours to see the hexadecimal mapping.
value = 0xB7
print('Hexadecimal:', format(value, '02X'))
print('Binary:', format(value, '08b'))
print('Groups:', format(value, '08b')[:4], format(value, '08b')[4:])
````

> **Common pitfall:** `1010`, `0b1010`, and `'1010'` are different. They are a decimal integer, a binary integer literal, and a string respectively.


## 3. Arithmetic Operators

Python's main arithmetic operators are:

| Operator | Meaning | Example | Result |
| --- | --- | --- | --- |
| `+` | addition | `7 + 3` | `10` |
| `-` | subtraction | `7 - 3` | `4` |
| `*` | multiplication | `7 * 3` | `21` |
| `/` | true division | `7 / 3` | `2.333...` |
| `//` | floor division | `7 // 3` | `2` |
| `%` | remainder | `7 % 3` | `1` |
| `**` | exponentiation | `7 ** 3` | `343` |


````python
a = 7
b = 3

print('addition:', a + b)
print('subtraction:', a - b)
print('multiplication:', a * b)
print('true division:', a / b)
print('floor division:', a // b)
print('remainder:', a % b)
print('exponentiation:', a ** b)
````

### Precedence and parentheses

Python broadly follows familiar arithmetic precedence:

1. parentheses
2. exponentiation
3. unary positive and negative
4. multiplication, division, floor division, and remainder
5. addition and subtraction

Operators at the same level usually evaluate left to right; exponentiation evaluates right to left. Use parentheses when they make the intended grouping clearer.


````python
print('2 + 3 * 4 =', 2 + 3 * 4)
print('(2 + 3) * 4 =', (2 + 3) * 4)
print('2 ** 3 ** 2 =', 2 ** 3 ** 2)
print('(2 ** 3) ** 2 =', (2 ** 3) ** 2)
````
