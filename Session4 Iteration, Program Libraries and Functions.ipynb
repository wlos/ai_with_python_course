## 1. Iteration: `while` and `for` Loops {#sec-iteration}

An **algorithm** is a finite sequence of steps for solving a problem. A useful way to plan a small algorithm is:

1. identify the input
2. decide what processing is required
3. produce the output

**Iteration** means repeating one or more steps. Each repetition is called an **iteration**.

| Loop | Use it when |
|:-----|:------------|
| `while` | repetition continues while a condition remains `True` |
| `for` | repetition processes each value supplied by an iterable |

: Choosing between Python loop statements

This section introduces `while` first, followed by `for`.


### Choosing the Loop {#sec-choosing-loop}

Ask these questions before writing a loop:

- Does repetition continue until a condition changes? Use `while`.
- Is the number of repetitions known? Use `for`. In this notebook, `range()` supplies the counter values.
- Which values must exist before the loop starts?
- Which values must change during each iteration?
- What result should be available after the loop finishes?


### 1.1 `while` Loops {#sec-while}

A `while` loop repeats while its condition is `True`.

``` {.python}
while condition:
    # repeated statements
    # update a value used by the condition
```

The algorithm must eventually make the condition `False`. Otherwise, the loop may never finish.


A repeated-addition algorithm needs an initial value, a condition, a repeated calculation, and an update. This preview adds the integers from 1 to 4.


````python
# Initialise the accumulator and the value tested by the while condition.
running_total = 0
current_number = 1

# Repeat while the current number is within the required range.
while current_number <= 4:
    # Add the current number to the running total.
    running_total += current_number
    # Move towards the stopping condition to avoid an infinite loop.
    current_number += 1

# Display the result after the loop has finished.
print('Total:', running_total)
````

This counter-controlled `while` loop displays the numbers 1 to 5. The update `number += 1` moves the algorithm towards its stopping condition.


#### Augmented Assignment Operators {#sec-augmented-assignment}

An **augmented assignment operator** combines a calculation with assignment. It updates a variable using its current value. These operators are common in loops because counters and accumulators must change during each iteration.

| Operator | Example | Equivalent statement |
|:---------|:--------|:---------------------|
| `+=` | `x += 2` | `x = x + 2` |
| `-=` | `x -= 2` | `x = x - 2` |
| `*=` | `x *= 2` | `x = x * 2` |
| `/=` | `x /= 2` | `x = x / 2` |
| `//=` | `x //= 2` | `x = x // 2` |
| `%=` | `x %= 2` | `x = x % 2` |
| `**=` | `x **= 2` | `x = x ** 2` |

: Common augmented assignment operators

::: {.callout-note}
`number += 1` increases `number` by one and stores the new value back in `number`. Remember that `/=` produces a floating-point result.
:::


````python
# Start the counter at the first number to display.
number = 1

# Continue while the counter is less than or equal to 5.
while number <= 5:
    print(number)
    # Increase the counter so the condition eventually becomes False.
    number += 1
````

A condition-controlled algorithm is useful when the number of iterations is not known in advance. This example counts how many weeks are needed to reach a savings target.


````python
# Define the starting balance, repeated deposit, target, and week counter.
savings_balance = 120
weekly_deposit = 55
savings_target = 300
weeks = 0

# Keep depositing until the balance reaches or exceeds the target.
while savings_balance < savings_target:
    savings_balance += weekly_deposit
    weeks += 1

# Report the final state after the condition becomes False.
print(f'Target reached after {weeks} weeks.')
print(f'Final balance: ${savings_balance}')
````

::: {.callout-warning}
## Common mistake: an infinite loop

In the following incorrect pattern, `count` never changes, so `count < 3` always remains `True`.

``` {.python}
count = 0
while count < 3:
    print(count)
    # Missing: count += 1
```

When designing a `while` loop, identify the update that will eventually make the condition `False`.
:::


### 1.2 `for` Loops {#sec-for}

A `for` loop repeats once for each value supplied by an iterable. In this notebook, `range()` supplies integer values and controls the number of repetitions.

``` {.python}
for counter in range(number_of_repetitions):
    # repeated statements
```

The loop variable receives the next integer during each iteration. Python uses indentation to show which statements belong to the loop.


Use `range()` when an algorithm needs a known number of repetitions. This example repeats the same message four times.


````python
# range(4) supplies four counter values: 0, 1, 2, and 3.
for repetition in range(4):
    # Use the loop counter in the message for the current iteration.
    print(f'Repetition {repetition}: practise iteration')
````

`range()` can use one, two, or three arguments.

| Pattern | Integers produced |
|:--------|:------------------|
| `range(stop)` | `0` up to, but not including, `stop` |
| `range(start, stop)` | `start` up to, but not including, `stop` |
| `range(start, stop, step)` | integers separated by `step`; `stop` is still excluded |

: Common forms of `range()`

::: {.callout-note}
The stop value is excluded. Remember this rule when checking for an off-by-one error.
:::


````python
# One argument starts at 0 and excludes the stop value 5.
print('range(5):')
for number in range(5):
    print(number, end=' ')

# Two arguments set start=1 and stop=6.
print()
print()
print('range(1, 6):')
for number in range(1, 6):
    print(number, end=' ')

# Three arguments add a step of 2.
print()
print()
print('range(2, 11, 2):')
for number in range(2, 11, 2):
    print(number, end=' ')
````

Set the start value in `range()` when numbering should begin at 1 rather than 0.


````python
# Start at 1 and stop before 4 to produce task numbers 1, 2, and 3.
for task_number in range(1, 4):
    print(f'Complete task {task_number}')
````

#### Counters and Accumulators {#sec-counters-accumulators}

A **counter** records how many times something happens. An **accumulator** builds a result such as a total. Both normally start before the loop and are updated inside the loop.


````python
# Initialise the counter and accumulator before the loop.
pass_count = 0
mark_total = 0

# Generate and process marks for five students.
for student_number in range(1, 6):
    mark = 35 + student_number * 12
    mark_total += mark

    # Count the mark only when it meets the pass condition.
    if mark >= 50:
        pass_count += 1

    print(f'Student {student_number}: {mark}')

# Use the accumulated values to report the class results.
print('Number of passes:', pass_count)
print('Average mark:', mark_total / 5)
````

A `for` loop can apply the same calculation to each integer produced by `range()`. This example converts four Celsius values without storing them in a collection.


````python
# range() supplies four Celsius values separated by a step of 3.
for temperature_c in range(18, 28, 3):
    # Convert the current value before displaying it.
    temperature_f = temperature_c * 9 / 5 + 32
    print(f'{temperature_c}°C = {temperature_f:.1f}°F')
````

## 2. Loop Control and Common Patterns {#sec-loop-control}

Two statements change the normal flow of a loop.

| Statement | Effect |
|:----------|:-------|
| `break` | end the nearest loop immediately |
| `continue` | skip the remaining statements in the current iteration |

: Loop-control statements

Use them when they make the algorithm clearer; too many early exits can make a loop difficult to trace.


Use `break` in a search algorithm when no further values need to be checked after the target is found.


````python
# Define the search target and record that it has not been found yet.
target_number = 7
found = False

# Check each candidate generated by range().
for candidate in range(1, 11):
    if candidate == target_number:
        found = True
        print('Target found:', candidate)
        # Stop immediately because no further search is required.
        break

# This branch runs only when the loop did not find the target.
if not found:
    print('Target was not found.')
````

Use `continue` to ignore one value and move directly to the next iteration.


````python
# Test each integer from 1 to 10.
for number in range(1, 11):
    if number % 3 == 0:
        # Skip the print operation for multiples of 3.
        continue

    print('Accepted number:', number)
````

### Nested Loops {#sec-nested-loops}

A nested loop places one loop inside another. The inner loop completes all its iterations for every single iteration of the outer loop. This is useful for grids and repeated combinations.


````python
# The outer loop controls each row.
for row in range(1, 4):
    # The inner loop completes all columns for the current row.
    for column in range(1, 4):
        print(f'{row} × {column} = {row * column}')

    # This statement runs once after each inner loop finishes.
    print('End of row')
````

### Tracing a Loop Algorithm {#sec-tracing-loops}

A trace table records the important variables after each iteration. For a loop over `range(1, 5)`, a running-total algorithm changes like this:

| Iteration | `number` | `total` after update |
|----------:|---------:|---------------------:|
| before loop | — | 0 |
| 1 | 1 | 1 |
| 2 | 2 | 3 |
| 3 | 3 | 6 |
| 4 | 4 | 10 |

: Trace table for a running-total algorithm

Trace tables help find incorrect initial values, missing updates, and off-by-one errors.


This algorithm calculates a total, a minimum, and a maximum for values generated from the loop counter. It does not store the generated values.


````python
# Initialise the total and use None until the first value is generated.
running_total = 0
smallest = None
largest = None

# Generate one value from each loop position.
for position in range(1, 6):
    value = 3 * position - 2
    running_total += value

    # Replace the minimum or maximum when the current value requires it.
    if smallest is None or value < smallest:
        smallest = value
    if largest is None or value > largest:
        largest = value

# Display the completed summary after all iterations.
print('Total:', running_total)
print('Smallest:', smallest)
print('Largest:', largest)
````

## 3. Program Libraries {#sec-program-libraries}

A program library provides reusable code. Python includes a large **standard library**, and additional third-party packages can be installed when required.

| Term | Meaning | Example |
|:-----|:--------|:--------|
| module | a Python file containing reusable code | `math`, `statistics`, `random` |
| function | named code that can receive values and return a result | `math.sqrt(81)` |
| class | a blueprint used to create objects | `random.Random` |
| object | one instance created from a class | `generator = random.Random(42)` |
| method | a function called through an object or class | `generator.randint(1, 10)` |

: Program-library terms

Imports make names from a module available to the current program.


Common import styles:

| Import statement | Example use | Comment |
|:-----------------|:------------|:--------|
| `import math` | `math.sqrt(81)` | keeps the module name visible |
| `from statistics import mean` | `mean(values)` | imports one selected name |
| `import random as rng` | `rng.randint(1, 10)` | uses an alias |
| `from module import *` | names used directly | generally avoid; name origins become unclear |

: Common Python import styles


Import the full `math` module when several of its constants or functions are required.


````python
# Import the complete module so its names use the math prefix.
import math

# Use the imported pi constant in the circle-area formula.
radius = 4
circle_area = math.pi * radius ** 2

# Call an imported function through the module name.
print(f'Area: {circle_area:.2f}')
print('Square root of 81:', math.sqrt(81))
````

Import selected functions when the short names remain clear in the program.


````python
# Import only the two statistics functions required in this cell.
from statistics import mean, median

# Store several numeric inputs in a Python list.
response_times = [12, 15, 14, 40, 13]

# Call each imported function directly with the list as its argument.
print('Mean:', mean(response_times))
print('Median:', median(response_times))
````

An alias provides a shorter local name. A fixed seed makes this teaching example reproducible, so it produces the same result each time the notebook runs.


````python
# Import random with a short local alias.
import random as rng

# Set a seed so the generated results are reproducible.
rng.seed(4)

# Generate and display five random integers.
for draw_number in range(1, 6):
    random_number = rng.randint(1, 10)
    print(f'Draw {draw_number}: {random_number}')
````

### Using a Library Class {#sec-library-class}

`Random` is a class from the `random` module. Calling the class creates an object. That object stores its own random-number state and provides methods such as `.randint()`.


````python
# Import the Random class directly from the random module.
from random import Random

# Create an object with its own reproducible random state.
generator = Random(42)

# Inspect the object type and call two of its methods.
print('Object type:', type(generator).__name__)
print('First random integer:', generator.randint(1, 10))
print('Second random integer:', generator.randint(1, 10))
````

### Creating and Using a Small Class {#sec-small-class}

You will often use classes supplied by libraries. You can also define a class when related data and behaviour belong together.

- `__init__` initialises a new object.
- `self` refers to the current object.
- attributes store object data.
- methods are functions defined inside the class.

Class names normally use `PascalCase`; function, method, and variable names normally use `snake_case`.


````python
# Define a class that groups study-session data and behaviour.
class StudySession:
    '''Store a topic and the number of minutes studied.'''

    # Initialise the attributes of each new object.
    def __init__(self, topic, minutes):
        self.topic = topic
        self.minutes = minutes

    # Return whether this object meets the supplied threshold.
    def is_long_session(self, threshold=60):
        return self.minutes >= threshold

    # Return a readable description of this object.
    def describe(self):
        return f'{self.topic}: {self.minutes} minutes'


# Create one object from the class.
session = StudySession('iteration', 75)

# Call methods with a default argument and a keyword argument.
print(session.describe())
print('Long session:', session.is_long_session())
print('At least 90 minutes:', session.is_long_session(threshold=90))
````

In `is_long_session`, `threshold` is a parameter. In `session.is_long_session(threshold=90)`, `90` is an argument. Methods use the same parameter, argument, and return concepts as other functions; the object is passed automatically as `self`.


## 4. Functions {#sec-functions}

A **function** is a named, reusable block of code that performs one focused task. Functions help organise algorithms, reduce repetition, and make programs easier to test.

``` {.python}
def function_name(parameters):
    '''Optional documentation string.'''
    # function body
    return result
```

Defining a function stores the instructions. Calling the function runs those instructions.


This function receives a name, builds a personalised message, and returns it to the caller.


````python
# Define a function with name as its parameter.
def create_greeting(name):
    '''Return a personalised greeting.'''
    # Build and return the result to the caller.
    message = f'Hello, {name}! Welcome to Session 4.'
    return message


# Pass an argument, store the returned value, and display it.
greeting = create_greeting('Aisha')
print(greeting)
````

### Parameters and Arguments {#sec-parameters-arguments}

| Term | Meaning | Example |
|:-----|:--------|:--------|
| parameter | a variable listed in the function definition | `name` in `def create_greeting(name):` |
| argument | a value supplied in a function call | `'Aisha'` in `create_greeting('Aisha')` |

: Parameters compared with arguments

Arguments can be passed by position or by keyword.


````python
# Define a function with two required parameters.
def calculate_rectangle_area(length, width):
    return length * width


# Call the same function with positional and keyword arguments.
positional_area = calculate_rectangle_area(8, 5)
keyword_area = calculate_rectangle_area(width=5, length=8)

# Both calling styles produce the same result.
print('Positional arguments:', positional_area)
print('Keyword arguments:', keyword_area)
````

### Default Parameters {#sec-default-parameters}

A default value makes a parameter optional. Required parameters must come before parameters with default values.


````python
# Define optional parameters with default values.
def calculate_total(price, quantity=1, discount_rate=0):
    # Calculate the subtotal before applying the discount.
    subtotal = price * quantity
    return subtotal * (1 - discount_rate)


# Compare calls that use defaults and keyword arguments.
print(calculate_total(12.50))
print(calculate_total(12.50, quantity=3))
print(calculate_total(12.50, quantity=3, discount_rate=0.10))
````

### Returning Values {#sec-return-values}

`return` sends a result back to the caller and ends the current function call. The caller can store, print, compare, or pass the returned value to another function.

A function without an explicit `return` returns `None`.


````python
# Return a calculated value so the caller can reuse it.
def square(number):
    return number ** 2


# Store one result and combine two other returned results.
squared_value = square(6)
combined_value = square(3) + square(4)

print('Stored result:', squared_value)
print('Combined returned results:', combined_value)
````

::: {.callout-warning}
## Common mistake: confusing `print` and `return`

`print()` displays text for a person. `return` gives a value back to the program. A printed value cannot automatically be reused in another calculation.
:::


````python
# This function displays a value but has no explicit return statement.
def display_double(number):
    print(number * 2)


# This function returns the calculated value to the caller.
def return_double(number):
    return number * 2


# Capture both function-call results for comparison.
displayed_result = display_double(5)
returned_result = return_double(5)

# A function without return produces None.
print('Value from display_double:', displayed_result)
print('Value from return_double:', returned_result)
````

A function can return multiple related values. Python packages them into a tuple, which can be unpacked into separate variables.


````python
# Calculate and return two related values from one function call.
def quotient_and_remainder(dividend, divisor):
    quotient = dividend // divisor
    remainder = dividend % divisor
    return quotient, remainder


# Unpack the returned tuple into two variables.
quotient, remainder = quotient_and_remainder(17, 5)

print('Quotient:', quotient)
print('Remainder:', remainder)
````

### Local Scope {#sec-local-scope}

Parameters and variables created inside a function are normally **local** to that function. Keeping temporary work local reduces unintended changes elsewhere in the program.


````python
# Define a global variable that can be read inside the function.
course_name = 'Introduction to Python'


def label_topic(topic):
    # label is local and exists only during this function call.
    label = f'{course_name}: {topic}'
    return label


# Call the function, then access the global variable directly.
print(label_topic('functions'))
print('Global variable:', course_name)
# The local variable label is not available outside label_topic().
````

### Functions Containing Loops {#sec-functions-loops}

A function can package a loop algorithm so it can be used with different arguments. This function counts multiples within an integer range.


````python
# Package a counting loop inside a reusable function.
def count_multiples(limit, divisor):
    multiple_count = 0

    # Test each integer from 1 through the inclusive limit.
    for number in range(1, limit + 1):
        if number % divisor == 0:
            multiple_count += 1

    # Return the completed count after the loop.
    return multiple_count


# Reuse the function with different divisor arguments.
print('Multiples of 3 from 1 to 20:', count_multiples(20, 3))
print('Multiples of 5 from 1 to 20:', count_multiples(20, 5))
````

### Functions Using Imported Library Components {#sec-functions-libraries}

Imports are usually placed near the top of a program or notebook. A function can then call the imported function whenever it runs.


````python
# Import the statistics function used by summarise_range().
from statistics import mean


def summarise_range(start, stop):
    '''Return key summary values for an integer range.'''
    # Create the range from the supplied arguments.
    number_range = range(start, stop)

    # Return several results together in a dictionary.
    return {
        'count': len(number_range),
        'mean': mean(number_range),
        'minimum': min(number_range),
        'maximum': max(number_range),
    }


# Store and display the returned summary.
range_summary = summarise_range(4, 10)
print(range_summary)
````

## 5. Common Pitfalls and Extensions {#sec-pitfalls}

| Pitfall | How to fix it |
|:--------|:--------------|
| incorrect indentation | use four spaces consistently for each code block |
| `range()` stops too early | remember that the stop value is excluded |
| a `while` loop never finishes | update a value used by the condition |
| accumulator starts with the wrong value | initialise totals and counts to `0` |
| `print()` is used when a result must be reused | return the value, then print it at the caller if needed |
| parameter and argument are confused | parameter is in the definition; argument is in the call |
| imported name is unclear | prefer `import module` or import only the names you need |
