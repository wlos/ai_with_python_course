## 1. What Are Artificial Intelligence and Machine Learning?

### 1.1 Artificial intelligence (AI)

**Artificial intelligence** is the broad field of building systems that perform tasks associated with intelligent behaviour, such as recognising objects, reasoning, planning, understanding language or choosing actions.

An AI system receives inputs and produces outputs. Its mechanism may use programmed rules, search, an expert system, machine learning, or a combination of these approaches. An ordinary calculation is not automatically AI just because it maps an input to an output.

![An AI system receives inputs and uses rules, expert knowledge or machine learning to produce an output.](assets/ai-input-output.svg)

An **expert system** combines a knowledge base with rules for drawing conclusions. A route-planning system can search possible routes without learning from historical trips.

Recall the Maqueen robot: an instruction such as “stop when the measured distance is below a threshold” follows a rule you wrote. It does not learn merely because it uses a sensor. Learning a model to recognise obstacles from examples would introduce ML.

### 1.2 Machine learning (ML)

**Machine learning** is a part of AI in which models learn patterns from examples. People still choose the task, data, algorithm and evaluation method, but the algorithm derives model parameters from training data rather than requiring every decision rule to be written by hand.

For example, an email classifier can learn patterns from messages labelled `spam` and `not spam`. It then uses those patterns to classify a new message.

![Historical input-output examples train model parameters, which are then used to make predictions for new inputs.](assets/ml-learning-from-history.svg)

- **Training:** use data to learn the model's parameters.
- **Inference:** use the trained model with a new input.
- **Evaluation:** compare model outputs with an appropriate measure of success.

### 1.3 Deep learning (DL)

**Deep learning** is a part of ML that uses neural networks with multiple layers of learnt transformations. These layers can learn representations: in an image model, early layers may detect edges, later layers may combine patterns into shapes, and further layers may support an object prediction.

Deep learning is widely used for images, speech and language. It can require substantial data and computation. Linear models and tree-based models remain useful, particularly for many tabular-data tasks.

### 1.4 How AI, ML and DL relate

![Nested areas show deep learning inside machine learning, and machine learning inside artificial intelligence.](assets/session11-ai-ml-dl.svg)

$$
\text{Deep learning} \subset \text{Machine learning} \subset \text{Artificial intelligence}
$$

| Area | Main idea | Example |
| --- | --- | --- |
| AI | Perform a task associated with intelligent behaviour | Search for a route; apply expert-system rules |
| ML | Learn useful patterns from data | Train a decision tree to classify emails |
| DL | Learn layered representations with a deep neural network | Train an image-recognition neural network |

All DL is ML; ML also includes methods that do not use deep networks. AI also includes methods that do not learn from data. A system can combine learnt predictions with programmed rules.

## 2. A short history of AI

The local course materials describe AI as a field with several waves of excitement, disappointment, and renewal. That is a helpful way to understand its history.

![A brief history of AI](assets/ai-history-timeline.png)

### Early ideas and the birth of AI

- In **1950**, Alan Turing asked whether machines could think, which helped shape the intellectual foundations of AI.
- In **1956**, the **Dartmouth Conference** gave the field its name: artificial intelligence.
- In the late 1950s, researchers such as **Frank Rosenblatt** introduced the **perceptron**, an early neural model.

### The first wave: symbolic AI

The first wave was strongly shaped by **symbolism**. Researchers believed intelligence could be expressed through symbols, logic, and rules. This led to systems that were good at formal reasoning, puzzle solving, and carefully defined tasks.

This approach was important, but it also had limits. Real life is messy. Human knowledge is full of exceptions, ambiguity, and common sense that are difficult to write down as neat logical rules.

### The second wave: expert systems

In the 1970s and 1980s, **expert systems** became popular. These systems stored large sets of hand-written rules from specialists in areas such as medicine or engineering. They could perform well in narrow domains, but they were expensive to build and hard to maintain.

### The third wave: machine learning and deep learning

From the late 20th century into the 21st century, AI shifted towards **learning from data** rather than relying only on hand-written rules. Several milestones helped this change:

- **1986**: backpropagation made neural network training far more practical;
- **1997**: IBM Deep Blue defeated chess champion Garry Kasparov;
- **2012**: AlexNet showed the power of deep learning for computer vision;
- **2017**: the Transformer architecture changed the direction of language and sequence modelling;
- **2022 onwards**: generative AI systems became widely visible in everyday life.

### Three classic schools of thought

The lecture slides also highlight three ways researchers have tried to build intelligence:

- **symbolism**: intelligence through logic and knowledge representation;
- **behaviourism**: intelligence through perception, action, and feedback control; and
- **connectionism**: intelligence through networks of neuron-like units.

Modern AI often combines ideas from more than one school, but deep learning is most closely linked to **connectionism**.

## 3. Fundamentals of Machine Learning

### 3.1 Data, algorithm and model

A **dataset** contains examples. **Features** represent the inputs. A **learning algorithm** uses training data to obtain a **model**, which contains the learnt relationships or parameters.

An algorithm is the procedure for learning; a trained model is the result of applying that procedure to data.

![Training data feeds a learning algorithm to create a model, which produces predictions from new data.](assets/machine-learning-basics.svg)

| Term | Meaning | Simple regression example |
| --- | --- | --- |
| Feature | Input used by the model | Delivery distance in kilometres |
| Target or label | Known result used in supervised learning | Recorded delivery time in minutes |
| Parameter | Value learnt during training | Slope and intercept of a fitted line |
| Hyperparameter | Setting selected before fitting and tuned using validation | Maximum depth of a decision tree |
| Loss | Quantity the training method seeks to reduce | Mean squared prediction error |
| Metric | Measure used to assess results | Mean absolute error in minutes |

A loss and an evaluation metric may be the same quantity, but they serve different purposes.

### 3.2 Supervised vs Unsupervised Learning

![Supervised learning uses labelled examples; unsupervised learning searches for structure without supplied targets.](assets/supervised-vs-unsupervised.svg)

| Aspect | Supervised learning | Unsupervised learning |
| --- | --- | --- |
| Training data | Features with known targets | Features without supplied targets |
| Learning objective | Learn a relationship between inputs and targets | Discover structure or patterns in the inputs |
| Common tasks | Regression and classification | Clustering and dimensionality reduction |
| Example | Predict delivery time from labelled past deliveries | Group deliveries with similar distances and parcel counts |

In **supervised learning**, known outcomes provide a training signal: a model learns to predict those outcomes for new examples. In **unsupervised learning**, no target labels are supplied to the learning algorithm; it searches for useful structure instead.

The same feature data can support different tasks. Delivery records with a known `late` target can train a supervised classifier. Using only the delivery features to discover groups is an unsupervised task.

### 3.3 Regression vs Classification vs Clustering

- **Regression** predicts a numeric quantity, such as delivery time or house price. Training uses examples with known numeric targets.
- **Classification** predicts one of the known categories, such as `late` or `on time`. Training uses examples with known class labels. Numeric labels such as `0` and `1` can represent categories.
- **Clustering** discovers groups of similar examples without supplied target labels. A cluster ID identifies a discovered group; it is not a predefined class or a quality ranking.

Regression and classification are supervised learning tasks; clustering is an unsupervised learning task.

![Regression predicts a number, classification predicts a known category, and clustering discovers groups.](assets/regression-classification-clustering-principle.svg)

## 4. Data for Machine Learning

### 4.1 Data for Machine Learning

Machine learning finds patterns in examples. A dataset normally contains observations in rows and attributes in columns.

| Term | Meaning | Example |
| --- | --- | --- |
| Observation, sample, or instance | One example represented by a row | one customer |
| Feature or data attribute | An input used to learn a pattern | monthly spend |
| Target or label | The result a supervised model learns to predict | whether a customer renewed |
| Identifier | A value used to distinguish records | customer ID |
| Feature matrix `X` | The selected input columns | age, spend, visits, channel |
| Target vector `y` | The selected target column | renewed |

An identifier is useful for checking and joining records, but it usually should not be a model feature. A model might memorise customer IDs instead of learning a useful relationship.

#### 4.1.1 Data Workflow

```text
[1] Load and inspect the data
    File or database -> DataFrame -> columns, types, and missing values
                    |
                    v
[2] Identify features, target, and identifiers
    DataFrame -> X feature matrix + y target vector
                    |
                    v
[3] Check data quality and possible bias
    Duplicates + missing values + group representation + target rates
                    |
                    v
[4] Clean values and plan transformations
    Clean -> impute -> standardise or scale -> encode
                    |
                    v
[5] Split labelled data into training and test sets
    train_test_split(X, y) -> X_train, X_test, y_train, y_test
```

### 4.2 Common Data Categories

#### 4.2.1 Structured, Semi-structured and Unstructured Data

| Category | Description | Examples |
| --- | --- | --- |
| Structured | Consistent rows, columns, and data types | CSV tables, Excel tables, database tables |
| Semi-structured | Keys or tags provide some organisation, but the shape may vary | JSON, XML, API responses |
| Unstructured | No ready-made table of observations and attributes | free-form text, images, audio, documents |

Unstructured data normally requires **feature extraction** before a conventional machine-learning model can use it. For example, text can be converted into word-frequency features.

#### 4.2.2 Labelled and Unlabelled Data

| Category | Meaning | Typical use |
| --- | --- | --- |
| Labelled | Each training observation has a known target | supervised learning |
| Unlabelled | Observations do not have a target | clustering and other unsupervised learning |

Structure and labelling describe different aspects of data. A table can be unlabelled, while a collection of unstructured images can be labelled with categories.

| Example | Structure | Labelling |
| --- | --- | --- |
| customer table with a `renewed` column | structured | labelled |
| customer table without an outcome column | structured | unlabelled |
| support messages with issue categories | unstructured | labelled |
| raw support messages only | unstructured | unlabelled |

#### 4.3.2 Load Structured Data from a Database

A relational database stores structured data in tables. SQLite is a lightweight database stored locally as one `.db` file. The example database is `sample_data/ml_training.db`, and its table is named `customer_training`.

A simple Python database workflow is:

1. connect to the database
2. write an SQL query
3. load the query result into a pandas `DataFrame`
4. close the connection

##### Direct SQLite Connection

A direct connection uses the connector for a particular database system. This example uses Python's built-in `sqlite3` connector. `contextlib.closing()` ensures the connection closes when the `with` block finishes.

````python
# Import sqlite3 to create a direct connection to the SQLite database.
import sqlite3
from contextlib import closing

DATABASE_PATH = 'sample_data/ml_training.db'

# pd.read_sql() runs the SQL query and returns the result as a DataFrame.
with closing(sqlite3.connect(DATABASE_PATH)) as connection:
    database_tables = pd.read_sql(
        "SELECT name FROM sqlite_master WHERE type = 'table' ORDER BY name",
        connection
    )

database_tables
````

After checking the available table names, use `SELECT *` to load every column and row from `customer_training` into a `DataFrame`.

````python
# sqlite3.connect() opens the database; closing() releases it after the query.
with closing(sqlite3.connect(DATABASE_PATH)) as connection:
    # pd.read_sql() loads every row and column returned by SELECT *.
    customers_from_database = pd.read_sql(
        'SELECT * FROM customer_training',
        connection
    )

# .shape checks the table size, and .head(3) previews three records.
print('Database shape:', customers_from_database.shape)
customers_from_database.head(3)
````

## 6. Glossary

| Term | Meaning |
| --- | --- |
| Artificial intelligence (AI) | The field of building systems that perform tasks associated with intelligent behaviour. |
| Machine learning (ML) | Methods that learn useful patterns from data. |
| Deep learning (DL) | ML using neural networks with multiple layers of learnt transformations. |
| Neural network | A model built from connected units that transform inputs using parameters and activation functions. |
| Observation or sample | One example in a dataset. |
| Feature | An input variable or representation used by a model. |
| Target or label | The known outcome used for supervised learning. |
| Identifier | A value used to distinguish or track records. |
| Algorithm | A procedure; in ML, a method used to learn a model from data. |
| Model | A learnt representation used to make predictions or describe patterns. |
| Parameter | An internal value learnt during training, such as a slope or weight. |
| Hyperparameter | A setting chosen before fitting, such as tree depth. |
| Training | Learning model parameters from training examples. |
| Inference | Applying a trained model to an input. |
| Loss | A quantity the learning process seeks to reduce. |
| Metric | A measure used to assess performance. |
| Validation set | Data used to compare modelling choices and tune settings. |
| Test set | Held-out data used for final evaluation of the selected model. |
| Generalisation | Useful performance on examples beyond the training data. |
| Overfitting | Learning training-specific details that transfer poorly to new data. |
| Underfitting | Failing to capture useful patterns because a model is too limited. |
| Supervised learning | Learning from examples with known targets. |
| Unsupervised learning | Learning structure without supplied target labels. |
| Reinforcement learning | Learning an action policy through interaction and rewards. |
| Regression | Predicting a numeric quantity. |
| Classification | Predicting a known category. |
| Clustering | Discovering groups based on similarity. |
| Preprocessing | Preparing data for a model, including imputation, scaling or encoding. |
| Data leakage | Using information during modelling that compromises an independent evaluation or would be unavailable at prediction time. |
| Data bias | A systematic distortion in data, measurements or labels. |
