## 1. Better Function Design {#sec-function-design}

In Session 4 you learnt how to define and call functions. This session focuses on designing functions that are easy to read, reuse, and share.

A well-designed function:

- does **one** focused task
- has a **clear, descriptive name** (use a verb, for example `calculate_average`, `check_temperature`)
- explains its purpose with a **docstring**
- accepts sensible **default values** when a parameter is optional
- **returns** a result rather than just printing it, so the caller can use the result

The table below compares a poorly designed function with a well-designed one.

| | Poorly designed | Well designed |
|:--|:--|:--|
| Name | `f` | `calculate_average` |
| Purpose unclear? | Yes | No — docstring explains it |
| Reusable result? | No — only prints | Yes — returns the value |
| Optional behaviour? | No | Yes — default parameter |


### 1.1 Docstrings {#sec-docstrings}

A **docstring** is a short description placed immediately inside a function using triple quotes `"""..."""`.
It explains what the function does, what its parameters mean, and what it returns.

Python can display a function's docstring at any time using `help()`.

``` python
def function_name(parameter):
    """One-line summary of what this function does.

    Args:
        parameter: description of the input.

    Returns:
        description of the output.
    """
    # function body
```


````python
# Example: a function with a docstring
def calculate_average(numbers):
    """Calculate and return the average of a list of numbers.

    Args:
        numbers: a list of numeric values.

    Returns:
        The average as a float, or 0 if the list is empty.
    """
    if len(numbers) == 0:   # guard against an empty list
        return 0
    return sum(numbers) / len(numbers)   # sum and len are built-in functions


scores = [72, 85, 91, 60, 78]
average = calculate_average(scores)
print(f"Average score: {average}")

# Display the docstring
help(calculate_average)
````

### 1.2 Default Parameter Values {#sec-defaults}

A **default value** is a fallback used when the caller does not supply that argument.
Default parameters must be listed **after** any parameters without defaults.

``` python
def greet(name, greeting="Hello"):   # greeting has a default value
    ...
```

This makes functions flexible: callers supply only what they need to change.


````python
# Default parameter: unit defaults to Celsius
def describe_temperature(temp, unit="C"):
    """Return a formatted temperature string with its unit.

    Args:
        temp: numeric temperature value.
        unit: temperature unit label, default is 'C' for Celsius.

    Returns:
        A string such as '36.6 degC'.
    """
    return f"{temp} deg{unit}"


# Call without the optional argument -- uses the default 'C'
print(describe_temperature(36.6))

# Call with a different unit
print(describe_temperature(97.9, "F"))
````

### 1.3 Type Hints {#sec-type-hints}

**Type hints** tell the reader what type of value each parameter and the return value should be.
Python does not enforce type hints at runtime -- they are a communication tool for programmers.

``` python
def add(a: int, b: int) -> int:
    ...
#         ^^^  parameter type    ^^^ return type
```

Common types: `int`, `float`, `str`, `bool`, `list`.


````python
# Type hints on parameters and return value
def celsius_to_fahrenheit(celsius: float) -> float:
    """Convert a Celsius temperature to Fahrenheit.

    Args:
        celsius: temperature in degrees Celsius.

    Returns:
        Equivalent temperature in degrees Fahrenheit.
    """
    # Standard conversion formula
    return celsius * 9 / 5 + 32


result = celsius_to_fahrenheit(100.0)
print(f"100 degC = {result} degF")
````

### 1.4 Returning Multiple Values {#sec-multiple-return}

A function can return more than one value by separating them with a comma.
Python packages them as a **tuple**, which you can unpack into separate variables.

``` python
low, high = find_range(data)   # unpack two returned values
```


````python
# Return two values: the minimum and maximum of a list
def find_range(numbers: list) -> tuple:
    """Return the minimum and maximum values in a list.

    Args:
        numbers: a non-empty list of numeric values.

    Returns:
        A tuple (minimum, maximum).
    """
    # min() and max() are built-in Python functions
    return min(numbers), max(numbers)


temperatures = [22.1, 18.4, 30.7, 25.0, 19.8]

# Unpack the returned tuple into two separate variables
low, high = find_range(temperatures)
print(f"Lowest:  {low} degC")
print(f"Highest: {high} degC")
````

## 2. What Is an Algorithm? {#sec-algorithms}

An **algorithm** is a finite, ordered sequence of steps that solves a problem.
Every algorithm has three parts:

| Part | Description | Example |
|:-----|:------------|:--------|
| **Input** | the data the algorithm starts with | a list of temperatures |
| **Process** | the steps applied to the input | find the largest value |
| **Output** | the result produced | the highest temperature |

You already use algorithms every day -- a recipe is an algorithm: ingredients (input), steps (process), meal (output).

When designing an algorithm, ask these questions before writing any code:

1. What is the input?
2. What steps are needed?
3. What is the expected output?
4. Are there any edge cases? (for example, an empty list or a negative number)


### 2.1 Describing an Algorithm in Plain Language {#sec-pseudocode}

Before writing Python, describe the algorithm in plain language -- sometimes called **pseudocode**.

**Example problem:** find the highest score in a list.

```
INPUT:  a list of scores
STEP 1: start with the first score as the current highest
STEP 2: for each remaining score:
            if it is greater than the current highest
                replace the current highest with this score
STEP 3: return the current highest
OUTPUT: the highest score
```

Python's built-in `max()` function does exactly this -- you do not need to write it yourself.
The pseudocode makes the *idea* visible before you reach for a tool.


````python
# max() does the work described in the pseudocode above
scores = [72, 85, 91, 60, 78]

# max() scans the list and returns the largest value
highest = max(scores)
print(f"Highest score: {highest}")

# min() works the same way for the smallest value
lowest = min(scores)
print(f"Lowest score:  {lowest}")
````

### 2.2 Tracing Code Step by Step {#sec-tracing}

**Tracing** means following a piece of code line by line and recording the value of each variable at each step.
It is the most reliable way to understand what code does -- and to find bugs.

The cell below contains a short function. Read through the trace table first, then run the cell to check your understanding.


````python
def count_above(numbers: list, threshold: float) -> int:
    """Count how many values in a list are strictly above a threshold.

    Args:
        numbers:   list of numeric values to check.
        threshold: the cut-off value; values must be strictly greater.

    Returns:
        Number of values that exceed the threshold.
    """
    count = 0               # initialise counter to zero
    for value in numbers:   # visit each element in the list
        if value > threshold:
            count += 1      # increment counter when the condition is met
    return count


temps = [18.0, 22.5, 30.1, 15.3, 27.8]
result = count_above(temps, 20.0)
print(f"Values above 20.0: {result}")
````

**Trace table** for `count_above([18.0, 22.5, 30.1, 15.3, 27.8], 20.0)`:

| Step | `value` | `value > 20.0` | `count` |
|-----:|--------:|:--------------|--------:|
| start | -- | -- | 0 |
| 1 | 18.0 | False | 0 |
| 2 | 22.5 | True | 1 |
| 3 | 30.1 | True | 2 |
| 4 | 15.3 | False | 2 |
| 5 | 27.8 | True | 3 |
| return | -- | -- | **3** |

The function returns `3`, which matches the printed output above.

## 3. Searching {#sec-searching}

**Searching** means finding a specific value (or its position) in a collection of data.

Python provides several built-in ways to search a list:

| Tool | What it does |
|:-----|:-------------|
| `value in my_list` | returns `True` if the value is present |
| `my_list.index(value)` | returns the position (index) of the first match |
| A `for` loop with `if` | lets you apply any condition while searching |


### 3.1 Linear Search {#sec-linear-search}

A **linear search** checks each element one by one from the start until it finds the value or reaches the end.
This is exactly what the `in` operator and `.index()` method do behind the scenes.

Linear search works on **any list** -- sorted or unsorted.


````python
# Using the 'in' operator -- the simplest way to check membership
subjects = ["Maths", "English", "Science", "History", "Art"]

search_term = "Science"

# 'in' returns True if the value appears anywhere in the list
if search_term in subjects:
    print(f"{search_term} is on the list.")
else:
    print(f"{search_term} was not found.")
````

````python
# Using .index() to find the position of a value
position = subjects.index("Science")
print(f"'Science' is at position {position} (0-indexed)")

# Safely search using 'in' before calling .index() to avoid a ValueError
target = "PE"
if target in subjects:
    print(f"'{target}' is at position {subjects.index(target)}")
else:
    print(f"'{target}' was not found in the list.")
````

### 3.2 Writing a Simple Search Function {#sec-search-function}

Sometimes you need to search with a **condition** rather than an exact match.
For example, find the first temperature that exceeds a threshold.
A short `for` loop with an `if` statement handles this neatly.


````python
def find_first_above(readings: list, threshold: float):
    """Return the index of the first value that exceeds a threshold.

    Args:
        readings:  list of numeric sensor readings.
        threshold: the value that a reading must exceed.

    Returns:
        The integer index of the first matching reading,
        or -1 if no reading exceeds the threshold.
    """
    for index, value in enumerate(readings):   # enumerate provides the index alongside each value
        if value > threshold:
            return index   # return immediately on the first match
    return -1              # -1 signals that no match was found


sensor_log = [21.0, 23.5, 25.1, 29.8, 31.4, 28.0]
alert_temp = 28.0

position = find_first_above(sensor_log, alert_temp)

if position != -1:
    print(f"First reading above {alert_temp} degC is at index {position}: {sensor_log[position]} degC")
else:
    print("No readings exceeded the threshold.")
````

### 3.3 Binary Search — the Guessing Game Idea {#sec-binary-search}

A **binary search** works on a **sorted** list. Instead of checking every element, it cuts the list in half each time:

1. Look at the **middle** element.
2. If it matches, you are done.
3. If the target is **smaller**, search the left half.
4. If the target is **larger**, search the right half.
5. Repeat until found or the search space is empty.

This is exactly the strategy used in the classic **guess the number** game.

Python's `bisect` module provides a ready-made binary search -- no need to write it from scratch.


````python
import bisect   # standard library module for binary search on sorted lists

# The list MUST be sorted for binary search to work correctly
sorted_scores = [55, 62, 70, 75, 80, 85, 91, 96]

target = 80

# bisect_left() returns the index where the target would be inserted
index = bisect.bisect_left(sorted_scores, target)

# Confirm the element at that index actually equals the target
if index < len(sorted_scores) and sorted_scores[index] == target:
    print(f"{target} found at index {index}")
else:
    print(f"{target} not found in the list")
````

### Micro:bit Experiment 4 — Guess the Number {.unnumbered}

``` python
from microbit import *

def play_guess_game(low: int, high: int) -> int:
    """Run a guess-the-number game and return the number of guesses taken.

    Think of a secret number between low and high.
    Press A if the displayed guess is too low.
    Press B if the displayed guess is too high.
    Press A and B together if the guess is correct.

    Args:
        low:  smallest possible secret number.
        high: largest possible secret number.

    Returns:
        The number of guesses taken to find the secret number.
    """
    guesses = 0

    while low <= high:
        mid = (low + high) // 2      # guess the midpoint of the current range
        display.scroll(str(mid), delay=80)
        guesses += 1

        # Wait until the player presses a button
        while not button_a.was_pressed() and not button_b.was_pressed():
            sleep(50)

        if button_a.is_pressed() and button_b.is_pressed():
            display.show(Image.HAPPY)   # both pressed: correct!
            break
        elif button_a.was_pressed():
            low = mid + 1             # secret is higher: search upper half
        else:
            high = mid - 1            # secret is lower: search lower half

    return guesses


total = play_guess_game(1, 16)
display.scroll(f"Done:{total}", delay=80)
```

## 4. Sorting {#sec-sorting}

**Sorting** means arranging data in a specific order -- usually ascending (smallest first) or descending (largest first).

Python provides two built-in tools:

| Tool | What it does | Modifies original? |
|:-----|:-------------|:------------------:|
| `sorted(my_list)` | returns a **new** sorted list | No |
| `my_list.sort()` | sorts the list **in place** | Yes |

Both accept an optional `key` argument to control how items are compared, and `reverse=True` to sort in descending order.


### 4.1 Basic Sorting {#sec-basic-sorting}


````python
scores = [72, 85, 91, 60, 78]

# sorted() returns a new list; the original is unchanged
ascending = sorted(scores)
print("Ascending: ", ascending)
print("Original unchanged:", scores)

# reverse=True sorts from largest to smallest
descending = sorted(scores, reverse=True)
print("Descending:", descending)
````

````python
# Sorting strings -- Python sorts alphabetically by default
names = ["Priya", "Alice", "Mohammed", "Chen", "Zara"]

# sorted() works on any iterable, not just numbers
sorted_names = sorted(names)
print("Alphabetical:", sorted_names)
````

### 4.2 Sorting with a Key {#sec-sort-key}

The `key` argument accepts a function that produces a comparison value for each element.
For example, `key=len` sorts strings by their length instead of alphabetically.


````python
# Sort subjects by name length (shortest first)
subjects = ["Mathematics", "Art", "Science", "PE", "English"]

# key=len tells sorted() to compare using the length of each string
by_length = sorted(subjects, key=len)
print("By length:", by_length)
````

````python
# Sort student records by score using a lambda function as the key
students = [
    {"name": "Alice",    "score": 85},
    {"name": "Mohammed", "score": 72},
    {"name": "Priya",    "score": 91},
    {"name": "Chen",     "score": 68},
]

# A lambda (anonymous function) extracts the 'score' value for comparison
ranked = sorted(students, key=lambda s: s["score"], reverse=True)

print("Ranking (highest first):")
for i, student in enumerate(ranked, start=1):   # start=1 begins the rank counter at 1
    print(f"  {i}. {student['name']}: {student['score']}")
````

### 4.3 How Sorting Works — a Visual Idea {#sec-sorting-intuition}

You do not need to implement a sorting algorithm yourself, but it helps to understand the core idea.

The simplest approach -- **bubble sort** -- repeatedly compares neighbouring pairs and swaps them if they are in the wrong order.

You can watch this process animated at [visualgo.net/en/sorting](https://visualgo.net/en/sorting).

The short example below shows **one pass** so you can see what a single round of comparisons and swaps looks like.


````python
# One pass of bubble sort: swap neighbours that are in the wrong order
data = [5, 3, 8, 1, 9, 2]
print("Before one pass:", data)

# Iterate through neighbouring pairs (positions i and i+1)
for i in range(len(data) - 1):
    if data[i] > data[i + 1]:
        # Swap using Python's simultaneous assignment
        data[i], data[i + 1] = data[i + 1], data[i]

print("After one pass: ", data)
print("(The largest value has moved to the end.)")

# In practice, always use Python's built-in sorted() for real code
print("\nFully sorted using sorted():", sorted([5, 3, 8, 1, 9, 2]))
````

## 5. Putting It All Together {#sec-together}

Good programs are built from small, focused functions that are **composed** -- each handles one step, and they connect to form the full solution.
This approach is called **modular design**.

The example below builds a small **grade analyser**:

```
INPUT:  a list of student score records
  |  calculate_average()   ->  average score
  |  sorted()              ->  ranked list
  |  find_failing()        ->  names of students below the pass mark
OUTPUT: a printed summary report
```


````python
# --- Helper functions ---

def calculate_average(numbers: list) -> float:
    """Return the average of a list of numbers, or 0 if the list is empty."""
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)


def find_failing(students: list, pass_mark: int = 50) -> list:
    """Return names of students whose score is below the pass mark.

    Args:
        students:  list of dicts with 'name' and 'score' keys.
        pass_mark: minimum score to pass; default is 50.

    Returns:
        A list of student names who did not reach the pass mark.
    """
    # List comprehension: collect names only where score is below pass_mark
    return [s["name"] for s in students if s["score"] < pass_mark]


def print_report(students: list, pass_mark: int = 50) -> None:
    """Print a formatted grade summary report.

    Args:
        students:  list of dicts with 'name' and 'score' keys.
        pass_mark: minimum score to pass; default is 50.
    """
    scores  = [s["score"] for s in students]   # extract scores into a plain list
    ranked  = sorted(students, key=lambda s: s["score"], reverse=True)
    avg     = calculate_average(scores)
    failing = find_failing(students, pass_mark)

    print("=== Grade Report ===")
    print(f"Class average: {avg:.1f}")
    print(f"Pass mark:     {pass_mark}")
    print()
    print("Rankings:")
    for i, s in enumerate(ranked, start=1):
        print(f"  {i}. {s['name']}: {s['score']}")
    print()
    if failing:
        print(f"Students below pass mark: {', '.join(failing)}")
    else:
        print("All students passed.")
````

````python
# --- Main program: define data and run the report ---

class_results = [
    {"name": "Alice",    "score": 85},
    {"name": "Mohammed", "score": 47},
    {"name": "Priya",    "score": 91},
    {"name": "Chen",     "score": 63},
    {"name": "Zara",     "score": 38},
]

# Call the top-level function -- it uses all helper functions internally
print_report(class_results)
````

## 6. Common Pitfalls {#sec-pitfalls}

| Pitfall | Example | How to fix it |
|:--------|:--------|:--------------|
| Calling `.index()` on a missing value raises `ValueError` | `[1, 2].index(5)` | Check `5 in my_list` first |
| Using `.sort()` when you need to keep the original | `.sort()` changes the list | Use `sorted(my_list)` instead |
| Sorting a mixed-type list raises `TypeError` | `sorted([1, 'a', 2])` | Ensure all elements are the same type |
| Forgetting that binary search requires a sorted list | Calling `bisect` on unsorted data | Always call `sorted()` before using `bisect` |
| Printing inside a function instead of returning | `print(result)` inside the function | Use `return result` and print at the call site |
